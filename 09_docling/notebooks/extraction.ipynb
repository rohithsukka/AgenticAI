```python
from pathlib import Path
import json

from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.document_converter import (
    DocumentConverter,
    PdfFormatOption,
)
```

```python
# -----------------------------
# Configuration
# -----------------------------

INPUT_PDF = "https://konverge.ai/pdf/Ebook-Agentic-AI.pdf"
OUTPUT_DIR = Path("output")

OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
#save the first 5 pages of PDF
import requests
from pypdf import PdfReader, PdfWriter
from io import BytesIO

# PDF URL
pdf_url = "https://konverge.ai/pdf/Ebook-Agentic-AI.pdf"

# Download the PDF
response = requests.get(pdf_url)
response.raise_for_status()

# Read the PDF from memory
reader = PdfReader(BytesIO(response.content))
writer = PdfWriter()

# Number of pages to extract
num_pages = min(5, len(reader.pages))

# Add the first 5 pages
for page_num in range(num_pages):
    writer.add_page(reader.pages[page_num])

# Save the new PDF
output_file = "Ebook-Agentic-AI_First_5_Pages.pdf"

with open(output_file, "wb") as f:
    writer.write(f)

print(f"Saved first {num_pages} pages as '{output_file}'")
```

```python
# -----------------------------
# Pipeline Options
# -----------------------------

pipeline_options = PdfPipelineOptions()

# OCR
pipeline_options.do_ocr = True

# Structured tables
pipeline_options.do_table_structure = True

# Generate page images
pipeline_options.generate_page_images = True

# Generate individual figure images
pipeline_options.generate_picture_images = True

# Better image quality
pipeline_options.images_scale = 2.0
```

```python
# -----------------------------
# Converter
# -----------------------------

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_options=pipeline_options
        )
    }
)
```

```python
# -----------------------------
# Convert
# -----------------------------
five_pages = "Ebook-Agentic-AI_First_5_Pages.pdf"
result = converter.convert(pdf_url)

document = result.document
```

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

headers_to_split_on = [
    ("#", "H1"),
    ("##", "H2"),
    ("###", "H3"),
]

markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on,
    strip_headers=False,   # Keep headings in the chunk
)

with open("./output/document_with_images.md", "r", encoding="utf-8") as f:
    markdown_text = f.read()

docs = markdown_splitter.split_text(markdown_text)

for i, doc in enumerate(docs):
    print("=" * 80)
    print(doc.metadata)
    print(doc.page_content[:500])
```
